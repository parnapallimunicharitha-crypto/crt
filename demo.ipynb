```python
import pandas as pd

data = {
    "players": ["virat", "Rohit", "sehwag", "dhoni", "virat", "Rohit", "sehwag", "dhoni"],
    "matches": [100, 200, 300, 400, 100, 200, 300, 400],
    "runs": [154, 200, 123, 257, 309, 198, 171, 120],
    "Balls": [50, 100, 120, 120, 150, 80, 120, 100]
}

ipl = pd.DataFrame(data)

print("Original Data:")
display(ipl)

# 1. Find strike rate of each player
ipl["Strike_Rate"] = (ipl["runs"] / ipl["Balls"]) * 100

print("1. Strike Rate of Each Player:")
display(ipl[["players", "runs", "Balls", "Strike_Rate"]])

# 2. Show only matches where Virat scored more than 150
virat_150 = ipl[(ipl["players"] == "virat") & (ipl["runs"] > 150)]

print("2. Virat scored more than 150:")
display(virat_150)

# 3. Show total runs scored by player and highest score by player
summary = ipl.groupby("players").agg(
    Total_Runs=("runs", "sum"),
    Highest_Score=("runs", "max")
)

print("3. Total Runs and Highest Score by Player:")
display(summary)
```
