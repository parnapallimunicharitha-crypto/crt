```python
import pandas as pd

data = {
    "players": ["virat", "Rohit", "sehwag", "dhoni", "virat", "Rohit", "sehwag", "dhoni"],
    "matches": [100, 200, 300, 400, 100, 200, 300, 400],
    "runs": [154, 200, 123, 257, 309, 198, 171, 120],
    "Balls": [50, 100, 120, 120, 150, 80, 120, 100]
}

ipl = pd.DataFrame(data)

print("Original Data:")
display(ipl)

# 1. Find strike rate of each player
ipl["Strike_Rate"] = (ipl["runs"] / ipl["Balls"]) * 100

print("1. Strike Rate of Each Player:")
display(ipl[["players", "runs", "Balls", "Strike_Rate"]])

# 2. Show only matches where Virat scored more than 150
virat_150 = ipl[(ipl["players"] == "virat") & (ipl["runs"] > 150)]

print("2. Virat scored more than 150:")
display(virat_150)

# 3. Show total runs scored by player and highest score by player
summary = ipl.groupby("players").agg(
    Total_Runs=("runs", "sum"),
    Highest_Score=("runs", "max")
)

print("3. Total Runs and Highest Score by Player:")
display(summary)
```

## median-median is the middle value when data is sorted 
#### 1,sort the data
#### 2,if odd count-middle value will be the median
#### 3, if even cout-avg of two middle val will be median




```python
marks=[110,70,86,65,70,80,90,100,5000] #skwed data or outlier data
mrks=[110,70,86,65,70,80,90,100] #removing outlier value for accurate mean calculation
def calculate_median(data):
    sorted_data = sorted(data)#ascending order
    n = len(sorted_data)
    if n % 2 == 0:
        return (sorted_data[n//2 - 1] + sorted_data[n//2]) / 2
    else:
        return sorted_data[n//2]
median_marks = calculate_median(marks)
print("Manual_Median Marks:", median_marks)
print("numpy median:", np.median(mrks))
#calculating median using Numpy
print("Median with outlier:",np.median(marks))
print("Median without outlier:",np.median(mrks))
```

### Mode
### mode is the most frequent value in dataset.its a measure of central tendency that wrks for categorical data.. 

```python
from scipy import stats
from collections import Counter

scores = [85, 90, 92, 85, 88, 95, 90, 85, 88]

def cal_mode(data):
    counter = Counter(data)
    maxdata = max(counter.values())
    modes = [k for k, v in counter.items() if v == maxdata]
    return modes, maxdata

manual_mode, frequency = cal_mode(scores)

print("Manual Mode:", manual_mode)
print("Frequency of Manual Mode:", frequency)
```
